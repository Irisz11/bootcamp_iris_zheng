```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from scipy.stats import skew, kurtosis
sns.set(context='talk', style='whitegrid')
np.random.seed(8)
pd.set_option('display.max_columns', 100)

# === Synthetic data generator (adapt or replace with your own data) ===
n = 160
df = pd.DataFrame({
    'date': pd.date_range('2021-02-01', periods=n, freq='D'),
    'region': np.random.choice(['North','South','East','West'], size=n),
    'age': np.random.normal(40, 8, size=n).clip(22, 70).round(1),
    'income': np.random.lognormal(mean=10.6, sigma=0.3, size=n).round(2),
    'transactions': np.random.poisson(lam=3, size=n),
})
base = df['income'] * 0.0015 + df['transactions']*18 + np.random.normal(0, 40, size=n)
df['spend'] = np.maximum(0, base).round(2)

# inject a bit of missingness and outliers
df.loc[np.random.choice(df.index, 5, replace=False), 'income'] = np.nan
df.loc[np.random.choice(df.index, 3, replace=False), 'spend'] = np.nan
df.loc[np.random.choice(df.index, 2, replace=False), 'transactions'] = df['transactions'].max()+12
df.head()
```

## 4b) Over time

Nothing above uses the `date` column — and a dataset with a date almost always has
something to say along it: a trend, a seasonal cycle, or a level shift where behaviour
changed.

This is not optional detail. Stage 09 builds lag and rolling features on exactly this
axis, and stage 10b splits the data on it. What you find here decides what those stages
can do — and if the rows are out of order, or there are gaps in the dates, both of
those stages produce nonsense without telling you.

```python
# TODO: plot at least one variable over time.
#   Hint: sort by `date` first, then set it as the index. Order matters — a rolling
#   mean computed over out-of-order rows is meaningless.
df_time = df.sort_values('date').set_index('date')

plt.figure(figsize=(10, 5))
plt.plot(df_time.index, df_time['spend'])
plt.title('Spend Over Time')
plt.xlabel('Date')
plt.ylabel('Spend')
plt.show()
# TODO: check for GAPS in the dates too. A rolling window that spans a missing week is
#   not the window you think it is.
date_diffs = df_time.index.to_series().diff()

print("Largest gap:", date_diffs.max())
print("Number of gaps greater than 1 day:", (date_diffs > pd.Timedelta(days=1)).sum())
# TODO: then say what you see in a markdown cell: trend, seasonality, a level shift, or
#   genuinely nothing. 'Nothing' is a real finding if you can show it.
```

### Time-Series Interpretation

**What?**  
The `spend` series fluctuates over time, with noticeable short-term variation but no clear long-term upward or downward trend.

**So what?**  
There does not appear to be strong seasonality or a major level shift in this synthetic dataset. This suggests that time itself may not be a dominant predictor of spending in the current sample.

**Now what?**  
For later stages, the data should remain sorted by `date` before creating lag or rolling features. Because there are no major date gaps, rolling-window features can be interpreted more reliably. Stage 09 can explore lagged and rolling versions of variables such as `spend`, while later train/test splits should preserve chronological order.

## Top 3 Insights

1. `income` is moderately right-skewed, with a skewness of about 0.99. This suggests that a log transformation may be worth considering before linear modeling if the skew affects model assumptions.

2. `transactions` contains clear extreme values and has very high skewness and kurtosis. These observations should be investigated before modeling because they may strongly influence fitted relationships.

3. `spend` has the strongest observed correlation with `transactions` (about 0.48), followed by `income` (about 0.31). `age` has almost no linear correlation with `spend`. These correlations are useful modeling clues but should not be interpreted as causal relationships.

## Assumptions & Risks

- The synthetic dataset is assumed to be representative enough to demonstrate the EDA workflow, but conclusions should not be generalized to real customer behavior.
- Missing values in `income` and `spend` must be handled before modeling. Any imputation method will introduce assumptions about why the values are missing.
- The extreme `transactions` values may be errors or valid rare observations. Removing them without investigation could discard useful information.
- Correlation measures linear association only and does not establish causality.
- The time-series plot does not show a strong trend, seasonality, or level shift, but the relatively short sample period may limit the ability to detect longer-term patterns.

## Implications for Next Step

Before modeling, I would:

- Handle missing values in `income` and `spend`.
- Investigate the extreme values in `transactions` and decide whether to retain, cap, or remove them based on context.
- Consider a log transformation for `income` because of its right-skewed distribution.
- Preserve chronological ordering before creating lag or rolling features.
- Explore `transactions` and `income` as potentially useful predictors of `spend`.
- Consider encoding `region` as a categorical feature because its categories are reasonably balanced.
